```python
df = pd.read_csv('dirty_transactions_dataset (1).csv')
print("Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nMissing values:\n", df.isnull().sum())
print("\nFirst 5 rows:\n", df.head())
```

```python
print("Unique countries BEFORE:", df['country'].unique())
df['country'] = df['country'].str.strip().str.title()
df['country'] = df['country'].replace({
    'Usa': 'United States',
    'Us': 'United States',
})
print("Unique countries AFTER:", df['country'].unique())
```

```python
print("Unique categories BEFORE:", df['product_category'].unique())
df['product_category'] = df['product_category'].str.strip().str.title()
print("Unique categories AFTER:", df['product_category'].unique())

print("\nUnique payment methods BEFORE:", df['payment_method'].unique())
df['payment_method'] = df['payment_method'].str.strip().str.title()
df['payment_method'] = df['payment_method'].fillna(df['payment_method'].mode()[0])
print("Unique payment methods AFTER:", df['payment_method'].unique())
```

```python
print("Unit price dtype BEFORE:", df['unit_price'].dtype)
print("Sample values BEFORE:", df['unit_price'].head(10).tolist())

df['unit_price'] = df['unit_price'].astype(str).str.replace(',', '')
df['unit_price'] = pd.to_numeric(df['unit_price'], errors='coerce')
df['unit_price'] = df['unit_price'].fillna(round(df['unit_price'].mean(), 2))

print("\nUnit price dtype AFTER:", df['unit_price'].dtype)
print("Missing unit prices remaining:", df['unit_price'].isnull().sum())

print("\nNegative quantities:", (df['quantity'] < 0).sum())
df = df[df['quantity'] > 0]
df['quantity'] = df['quantity'].fillna(df['quantity'].median())
print("Missing quantities remaining:", df['quantity'].isnull().sum())
print("Shape after removing negatives:", df.shape)
```

```python
print("Missing ages BEFORE:", df['age'].isnull().sum())
df['age'] = df['age'].fillna(round(df['age'].mean())).astype(int)
print("Missing ages AFTER:", df['age'].isnull().sum())

print("\nMissing dates BEFORE:", df['transaction_date'].isnull().sum())
df = df.dropna(subset=['transaction_date'])
df['transaction_date'] = pd.to_datetime(df['transaction_date'])
print("Missing dates AFTER:", df['transaction_date'].isnull().sum())
print("Date dtype:", df['transaction_date'].dtype)

df['revenue'] = df['unit_price'] * df['quantity']
print("\nRevenue column created!")
print("Sample revenue values:", df['revenue'].head().tolist())

print("\n========== FINAL SUMMARY ==========")
print("Final Shape:", df.shape)
print("\nRemaining Missing Values:")
print(df.isnull().sum())
print("\nData Types:")
print(df.dtypes)
```

```python
df.to_csv('cleaned_fmcg_dataset.csv', index=False)
print("Cleaned dataset saved!")

from google.colab import files
files.download('cleaned_fmcg_dataset.csv')
```
